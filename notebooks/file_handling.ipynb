# Python File Handling: Complete Guide

File handling is one of the most important concepts in Python. It allows you to:
- **Read** files (extract data)
- **Write** files (save data)
- **Append** files (add data)
- **Create** and **Delete** files

This notebook will cover everything you need to know about working with files in Python.

## 1. Basic File Operations

### The open() Function
The `open()` function is the gateway to file handling in Python.

**Syntax:**
```python
file = open(filename, mode)
```

**File Modes:**
- `'r'` - **Read** (default): Opens a file for reading. Error if file doesn't exist
- `'a'` - **Append**: Opens a file for appending. Creates file if it doesn't exist
- `'w'` - **Write**: Opens a file for writing. Creates the file or overwrites existing content
- `'x'` - **Create**: Creates a file. Error if file already exists
- `'b'` - **Binary**: Used with other modes (e.g., 'rb', 'wb')
- `'t'` - **Text**: Default mode (e.g., 'rt', 'wt')
- `'+'` - **Read and Write**: Combined with other modes (e.g., 'r+', 'w+')

````python
file= open("test.txt","w")
file.write("This file is created by Arvind\n")
file.write("This is the second line\n")
file.close()
print("File created successfully")

import os
print(f"file created at: {os.getcwd()}/test.txt")
print(f"file exists: {os.path.exists('test.txt')}")
print(f"file size: {os.path.getsize('test.txt')} bytes")
````

````python
f = open("test.txt","r")
content = f.read()
print("File content:")
print(content)
f.close()

f=open("test.txt","r")
line1 = f.readline()
line2 = f.readline()
print("First line:")
print(line1)
print("Second line:")
print(line2)
f.close()
````

## 2. The `with` Statement (Context Manager) - BEST PRACTICE ⭐

**Why use `with`?**
- Automatically closes the file (even if an error occurs)
- Cleaner and more readable code
- Prevents resource leaks

**Syntax:**
```python
with open(filename, mode) as file:
    # Work with file
    # File automatically closes after this block
```

````python
with open("test.txt","a+") as f:
    f.write("\nThis is the new fourth line added\n")
````

````python
with open("test.txt","r") as f:
    print(f.read())
````

## 3. Different File Modes Explained

### Write Mode ('w') - Overwrites entire file
Creates a new file or **overwrites** existing content (data loss risk!)

### Append Mode ('a') - Adds to end of file
Safely adds new data without losing existing content

### Create Mode ('x') - Creates new file only
Fails if file already exists (prevents accidental overwrite)

````python
### Example 4B: Append Mode ('a') - Safe Addition

# Create initial file
with open('log.txt', 'w') as file:
    file.write("Log Entry 1: System started\n")

with open('log.txt', 'r') as file:
    print("Initial content:")
    print(file.read())

print("-" * 40)

# Append new data (old data is preserved!)
with open('log.txt', 'a') as file:
    file.write("Log Entry 2: User logged in\n")
    file.write("Log Entry 3: File processed\n")

with open('log.txt', 'r') as file:
    print("\nAfter appending (all entries preserved!):")
    print(file.read())
````

````python
try:
    with open("unique_file.txt", 'x') as file:
        file.write("this is a unique file created with 'x' mode\n")
    print("unique_file.txt created successfully.")
except FileExistsError:
    print("Error: unique_file.txt already exists. 'x' mode prevents overwriting.")

try:
    with open("unique_file.txt", 'x') as file:
        file.write("Trying again to create a unique file.\n")
    print("unique_file.txt created successfully.")
except FileExistsError:
    print("Error: unique_file.txt already exists.This prevents data loss by not allowing")
````

## 4. Working with CSV Files

CSV (Comma-Separated Values) files are commonly used for tabular data. Python's `csv` module makes it easy to work with them.

````python
### Example 5: Reading and Writing CSV Files

import csv

# Create a CSV file with student data
data = [
    ['Name', 'Age', 'Grade'],
    ['Alice', '20', 'A'],
    ['Bob', '19', 'B'],
    ['Charlie', '21', 'A'],
    ['Diana', '20', 'A+']
]

# Write CSV
with open('students.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)
print("✓ CSV file created\n")

# Read CSV - Method 1: Using csv.reader
print("Method 1: Using csv.reader()")
print("-" * 40)
with open('students.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(f"{row[0]:10} | Age: {row[1]:3} | Grade: {row[2]}")

print("\n" + "=" * 40 + "\n")

# Read CSV - Method 2: Using csv.DictReader (more convenient)
print("Method 2: Using csv.DictReader()")
print("-" * 40)
with open('students.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(f"{row['Name']:10} | Age: {row['Age']:3} | Grade: {row['Grade']}")
````

## 5. Working with JSON Files

JSON (JavaScript Object Notation) is perfect for storing structured data like dictionaries and lists. Python's `json` module handles conversion automatically.

````python
### Example 6: Working with JSON Files

import json

# Create and write JSON data
user_data = {
    'name': 'John Doe',
    'email': 'john@example.com',
    'age': 30,
    'hobbies': ['reading', 'gaming', 'coding'],
    'address': {
        'street': '123 Main St',
        'city': 'New York',
        'country': 'USA'
    }
}

# Write JSON
with open('user.json', 'w') as file:
    json.dump(user_data, file, indent=4)  # indent for pretty printing
print("✓ JSON file created\n")

# Read JSON
with open('user.json', 'r') as file:
    loaded_data = json.load(file)

print("Loaded JSON data:")
print(f"Name: {loaded_data['name']}")
print(f"Email: {loaded_data['email']}")
print(f"Age: {loaded_data['age']}")
print(f"Hobbies: {', '.join(loaded_data['hobbies'])}")
print(f"City: {loaded_data['address']['city']}")

print("\n" + "-" * 40)
print("\nPretty-printed JSON:")
print(json.dumps(loaded_data, indent=4))
````

## 6. Modern File Operations with `pathlib` Module

The `pathlib` module provides an object-oriented approach to file paths - cleaner and cross-platform compatible.

````python
### Example 7: Using pathlib for Modern File Handling

from pathlib import Path

# Create a Path object
file_path = Path('pathlib_demo.txt')

# Write to file using pathlib
file_path.write_text("This is created with pathlib!\nMuch cleaner syntax!")
print("✓ File created with pathlib\n")

# Read from file
content = file_path.read_text()
print("Content:")
print(content)

print("\n" + "-" * 40 + "\n")

# Check file properties
print(f"File name: {file_path.name}")
print(f"File stem (without extension): {file_path.stem}")
print(f"File suffix: {file_path.suffix}")
print(f"Absolute path: {file_path.absolute()}")
print(f"Parent directory: {file_path.parent}")
print(f"File exists: {file_path.exists()}")
print(f"Is file: {file_path.is_file()}")
print(f"File size: {file_path.stat().st_size} bytes")

print("\n" + "-" * 40 + "\n")

# List all files in current directory
print("Python files in current directory:")
for py_file in Path('.').glob('*.py'):
    print(f"  - {py_file.name}")

print("\nText files in current directory:")
for txt_file in Path('.').glob('*.txt'):
    print(f"  - {txt_file.name}")
````

````python
### Scenario 3: Batch Processing (Converting File Formats)

print("\n" + "=" * 60)
print("SCENARIO 3: BATCH PROCESSING (CSV → JSON)")
print("=" * 60)

import csv
import json

# Create sample CSV data
csv_content = """product_id,product_name,price,quantity
1,Laptop,999.99,5
2,Mouse,25.50,50
3,Keyboard,79.99,30
4,Monitor,349.99,10"""

# Write CSV file
with open('products.csv', 'w') as file:
    file.write(csv_content)

# Convert CSV to JSON
products = []

with open('products.csv', 'r') as csv_file:
    reader = csv.DictReader(csv_file)
    for row in reader:
        # Convert string numbers to proper types
        product = {
            'id': int(row['product_id']),
            'name': row['product_name'],
            'price': float(row['price']),
            'quantity': int(row['quantity'])
        }
        products.append(product)

# Save as JSON
with open('products.json', 'w') as json_file:
    json.dump(products, json_file, indent=2)

print("✓ Converted CSV to JSON\n")

# Display the results
print("Original CSV content:")
print(csv_content)

print("\n" + "-" * 60)
print("\nConverted JSON format:")
with open('products.json', 'r') as f:
    print(f.read())
````

````python
### Scenario 4: Creating Backups (Append Mode Use Case)

print("\n" + "=" * 60)
print("SCENARIO 4: BACKUP SYSTEM (Daily User Actions Log)")
print("=" * 60)

from datetime import datetime

# Simulate user actions over multiple days
def log_user_action(action, user="john_doe"):
    """Append action to backup log file"""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] User: {user} | Action: {action}\n"
    
    # Use 'a' mode to append safely without overwriting
    with open('user_actions_backup.log', 'a') as file:
        file.write(log_entry)

# Simulate various user actions
print("Logging user actions...")
log_user_action("Login")
log_user_action("Created new project")
log_user_action("Updated profile")
log_user_action("Downloaded report")
log_user_action("Logout")

print("✓ Actions logged to backup file\n")

# Read and display all logged actions
print("Backup Log Contents:")
print("-" * 60)
with open('user_actions_backup.log', 'r') as file:
    content = file.read()
    if content:
        print(content)
    else:
        print("(log file is empty)")

# Count total actions
with open('user_actions_backup.log', 'r') as file:
    action_count = len(file.readlines())

print(f"Total logged actions: {action_count}")
````

````python
### Scenario 5: Error Handling in File Operations

print("\n" + "=" * 60)
print("SCENARIO 5: ROBUST ERROR HANDLING")
print("=" * 60)

def safe_read_file(filename):
    """Read file safely with error handling"""
    try:
        with open(filename, 'r') as file:
            return file.read()
    except FileNotFoundError:
        print(f"❌ Error: File '{filename}' not found")
        return None
    except PermissionError:
        print(f"❌ Error: Permission denied to access '{filename}'")
        return None
    except Exception as e:
        print(f"❌ Unexpected error: {e}")
        return None

def safe_write_file(filename, content):
    """Write file safely with error handling"""
    try:
        with open(filename, 'w') as file:
            file.write(content)
        print(f"✓ Successfully wrote to '{filename}'")
        return True
    except PermissionError:
        print(f"❌ Error: Permission denied to write to '{filename}'")
        return False
    except IOError as e:
        print(f"❌ IO Error: {e}")
        return False

# Test: Try to read non-existent file
print("\nTest 1: Reading non-existent file")
result = safe_read_file('non_existent_file.txt')

# Test: Create and write file
print("\nTest 2: Writing to file")
safe_write_file('protected_data.txt', 'Important data: 12345')

# Test: Read the file we just created
print("\nTest 3: Reading created file")
content = safe_read_file('protected_data.txt')
if content:
    print(f"File contents: {content}")
````

## 8. Best Practices & Common Mistakes

### ✅ DO's:
| Practice | Why | Example |
|----------|-----|---------|
| Use `with` statement | Auto-closes files, cleaner code | `with open(...) as f:` |
| Use `pathlib` for paths | Cross-platform, object-oriented | `from pathlib import Path` |
| Use 'a' mode to preserve data | Prevents accidental data loss | `open(file, 'a')` for logs |
| Handle exceptions | Prevents crashes | `try/except FileNotFoundError` |
| Use context managers | Ensures cleanup | `with` block automatically closes |

### ❌ DON'Ts:
| Mistake | Problem | Better Way |
|---------|---------|-----------|
| Forget to `.close()` | Resource leaks | Use `with` statement |
| Use 'w' when you mean 'a' | Data loss! | Use 'a' to append safely |
| No error handling | Crashes silently | Use try/except blocks |
| Hard-code paths | Not portable | Use `pathlib` or relative paths |
| Read huge files all at once | Memory problems | Use line-by-line or chunks |

## 9. Quick Reference Guide

### Common File Operations Cheat Sheet

```python
# 📖 READING FILES
with open('file.txt', 'r') as f:
    content = f.read()           # Read entire file
    lines = f.readlines()        # Read as list of lines
    f.seek(0)                    # Go back to start

# ✍️  WRITING FILES
with open('file.txt', 'w') as f:
    f.write('text')              # Overwrite entire file
    f.writelines(lines)          # Write list of lines

# ➕ APPENDING FILES
with open('file.txt', 'a') as f:
    f.write('append this')       # Add without overwriting

# 📊 WORKING WITH CSV
import csv
with open('file.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)

# 📋 WORKING WITH JSON
import json
data = json.load(open('file.json'))
json.dump(data, open('file.json', 'w'), indent=2)

# 🛣️  USING PATHLIB
from pathlib import Path
p = Path('file.txt')
content = p.read_text()
p.write_text('new content')
```
