```python
def even_or_odd(num):
    """This will check even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
```

```python
### Variable length arguments  
## Understanding *args and **kwargs
def sum_numbers(*args):
    """This will add all the given numbers"""
    return sum(args)


def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
sum_numbers(100,101,102,103,104,105,8)
```

```python
print_details(name="Aravind", age=27, city="Hyderabad")
```

```python
def details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}")

details(name="Aravind", age=32, city="Hyderabad")
```

```python
def Student_info(*args,**kwargs):
    print("Subjects:", args)
    print("Information:", kwargs)

Student_info("Maths", "Science", "Social", name="Aravind", age=24, city="Hyderabad")
```

**Temperature Conversion**

```python
def convert_temp(temp,unit):
    """This will do temperature conversion"""
    if unit=='C':
        return temp * 9/5 + 32  ## Celsius to Fahrenheit
    elif unit=='F':
        return (temp-32)*5/9    ## Fahrenheit to Celsius 
    else:
        return None

print(convert_temp(32,'C'))
print(convert_temp(104,'F'))
```
