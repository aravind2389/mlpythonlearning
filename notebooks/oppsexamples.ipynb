```python
# CLASS = Blueprint for creating bank accounts
class BankAccount:

    # __init__ is a special method (constructor)
    # It runs automatically when we create an object
    def __init__(self, name, balance):
        self.name = name        # Each attribute will have its own name
        self.balance = balance  # Each attribute will have its own balance

    # Method to deposit money
    def deposit(self, amount):
        self.balance += amount  # Add amount to balance
        print(f"{self.name} deposited {amount}")
        print(f"New balance: {self.balance}")

    # Method to withdraw money
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance ❌")
        else:
            self.balance -= amount
            print(f"{self.name} withdrew {amount}")
            print(f"Remaining balance: {self.balance}")

    # Method to check balance
    def check_balance(self):
        print(f"{self.name}'s current balance is {self.balance}")


# OBJECTS = Real bank accounts created from the class

# Creating first account
acc1 = BankAccount("Ravi", 1000)

# Creating second account
acc2 = BankAccount("Anita", 2000)


# USING OBJECTS

acc1.check_balance()
acc2.check_balance()

# Ravi deposits money
acc1.deposit(500)

# Anita withdraws money
acc2.withdraw(900)
acc2.withdraw(1100) 

# Check balances
acc1.check_balance()
acc2.check_balance()
```

```python
class ShoppingCart:
    def __init__(self, owner, items):
        self.owner = owner  # Owner of the shopping cart
        self.items = list((items))   # List of items in the cart

#Adding items to the cart
    def add_item(self, item):
        self.items.append(item)  # Add item to the cart
        print(f"Added {item} to the cart.")

#Removing an item from the cart
    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)  # Remove item from the cart
            print(f"Removed {item} from the cart.")
        else:
            print(f"{item} not found in the cart.")

    def view_cart(self):
        if not self.items:
            print("Your cart is empty.")
        else:
            print("Items in your cart:")
            for item in self.items:
                print(f"- {item}")


# Creating a shopping cart for Alice
cart1 = ShoppingCart("Alice", ["Apples", "Bananas"])

cart2 = ShoppingCart("Bob", ["Oranges", "Grapes"])

cart1.view_cart()  # View items in the cart
cart2.view_cart()  # View items in the cart

cart1.add_item("Milk")  # Add an item to Alice's cart
cart1.view_cart() # View updated cart for Alice

cart1.remove_item("Bananas")  # Remove an item from Alice's cart
cart1.view_cart() # View updated cart for Alice

cart1.remove_item("Bananas")
```

**Step-by-step thinking process (use this for ANY problem)**

Whenever you get a requirement, don’t jump into code. Break it like this:

**1. Identify “things” (→ Classes)**

Ask:

What are the main entities in this system?

For your e-commerce example:

Product
User
Cart
Order

👉 These become classes

**2. Identify “data” (→ Attributes)**

Ask:

What information does each thing store?

Example:

Product → name, price
User → name, cart, orders
Cart → items

👉 These become attributes

**3. Identify “actions” (→ Methods)**

Ask:

What can this thing DO?

Example:

Cart → add_item(), remove_item()
User → checkout()
Order → (mostly stores data)

👉 These become methods

**4. Identify relationships**

Ask:

How are these connected?

Example:

User has a Cart
Cart has many Products
User has many Orders

👉 This helps structure your code correctly

```python
# CLASS = Product → blueprint for products
class Product:

    def __init__(self, name, price):
        # ATTRIBUTES (data of each product object)
        self.name = name     # product name (e.g., Laptop)
        self.price = price   # product price (e.g., 50000)


# CLASS = ShoppingCart → blueprint for cart
class ShoppingCart:

    def __init__(self):
        # ATTRIBUTE → each cart has its own list of items
        self.items = []  
        # Example structure:
        # [{"product": product_object, "quantity": 2}]

    # METHOD → action (add item to cart)
    def add_item(self, product, quantity):

        # loop through existing items (THIS cart's data)
        for item in self.items:

            # check if product already exists
            if item["product"].name == product.name:

                # update quantity (modify attribute data)
                item["quantity"] += quantity
                print(f"Updated {product.name} quantity to {item['quantity']}")
                return

        # if product not found → add new item
        self.items.append({"product": product, "quantity": quantity})
        print(f"{product.name} x{quantity} added to cart")

    # METHOD → remove item
    def remove_item(self, product_name):
        for item in self.items:
            if item["product"].name == product_name:
                self.items.remove(item)
                print(f"{product_name} removed from cart")
                return
        print(f"{product_name} not found")

    # METHOD → update quantity
    def update_quantity(self, product_name, quantity):
        for item in self.items:
            if item["product"].name == product_name:

                # change attribute (quantity)
                item["quantity"] = quantity
                print(f"{product_name} quantity updated to {quantity}")
                return
        print(f"{product_name} not found")

    # METHOD → display cart
    def view_cart(self):
        print("\nCart items:")

        if not self.items:
            print("Cart is empty")
            return

        for item in self.items:
            p = item["product"]   # product object
            q = item["quantity"]  # quantity value

            # accessing product attributes
            print(f"- {p.name} | Price: {p.price} | Qty: {q}")

    # METHOD → calculate total price
    def calculate_total(self):
        total = 0

        # loop through cart items
        for item in self.items:
            # using product attributes + quantity
            total += item["product"].price * item["quantity"]

        return total


# CLASS = Order → represents a completed purchase
class Order:

    def __init__(self, items, total):
        # ATTRIBUTES
        self.items = items   # list of items purchased
        self.total = total   # total amount


# CLASS = User → represents a customer
class User:

    def __init__(self, name):
        # ATTRIBUTES
        self.name = name                 # user name
        self.cart = ShoppingCart()       # each user HAS a cart (object)
        self.orders = []                 # list to store past orders

    # METHOD → add product to cart
    def add_to_cart(self, product, quantity):
        print(f"\n{self.name} adding items...")

        # calling another object's method
        self.cart.add_item(product, quantity)

    # METHOD → apply discount
    def apply_discount(self, total):
        if total > 50000:
            print("Applying 10% discount")
            return total * 0.9
        return total

    # METHOD → checkout process
    def checkout(self):
        print(f"\n{self.name} is checking out...")

        # get total from cart object
        total = self.cart.calculate_total()

        # apply discount
        total = self.apply_discount(total)

        print(f"Final amount: {total}")

        # create ORDER OBJECT
        order = Order(self.cart.items.copy(), total)

        # store in user's order history
        self.orders.append(order)

        # clear cart after checkout
        self.cart.items.clear()

    # METHOD → view past orders
    def view_orders(self):
        print(f"\n{self.name}'s Orders:")

        if not self.orders:
            print("No orders yet")
            return

        for i, order in enumerate(self.orders, 1):
            print(f"\nOrder {i} - Total: {order.total}")

            for item in order.items:
                print(f"- {item['product'].name} x{item['quantity']}")


# ------------------ OBJECTS (REAL INSTANCES) ------------------

# Product objects
p1 = Product("Laptop", 50000)
p2 = Product("Phone", 20000)
p3 = Product("Headphones", 2000)

# User object
user = User("Ravi")


# ------------------ USING OBJECTS ------------------

user.add_to_cart(p1, 1)
user.add_to_cart(p3, 2)

user.add_to_cart(p3, 1)   # updates quantity

user.cart.view_cart()

user.cart.update_quantity("Laptop", 2)

user.cart.remove_item("Headphones")

user.cart.view_cart()

user.checkout()

user.view_orders()
```

**Inheritance and its types**

```python
#Account class with inheritance and additional features
class Account:  # Base class (parent) - defines common attributes and methods for all account types
    def __init__(self, acc_no, name, balance):  # Constructor for base class - initializes shared attributes
        self.acc_no = acc_no  # Instance attribute: account number (inherited by subclasses)
        self.name = name      # Instance attribute: account holder name (inherited by subclasses)
        self.balance = balance  # Instance attribute: current balance (inherited by subclasses)

    def deposit(self, amount):  # Method in base class - can be inherited or overridden
        self.balance += amount  # Updates balance attribute
        print(f"{self.name} deposited {amount}. New balance: {self.balance}")  # Prints deposit info

    def withdraw(self, amount):  # Method in base class - can be overridden by subclasses
        if amount > self.balance:  # Checks if withdrawal exceeds balance
            print("Insufficient balance")  # Error message if insufficient funds
        else:
            self.balance -= amount  # Deducts amount from balance
            print(f"{self.name} withdrew {amount}. Remaining balance: {self.balance}")  # Prints withdrawal info

    def check_balance(self):  # Method in base class - inherited by subclasses
        print(f"{self.name}'s current balance is {self.balance}")  # Prints current balance
        

#Current account with overdraft feature
class CurrentAccount(Account):  # Subclass (child) inherits from Account (hierarchical inheritance)
    def __init__(self, acc_no, name, balance=0, overdraft_limit=10000):  # Overrides __init__ to add overdraft_limit
        super().__init__(acc_no, name, balance)  # Calls parent __init__ to set inherited attributes (acc_no, name, balance)
        self.overdraft_limit = overdraft_limit  # Adds new attribute specific to CurrentAccount

    def withdraw(self, amount):  # Overrides parent withdraw method (polymorphism)
        available_funds = self.balance + self.overdraft_limit  # Calculates total available funds (uses inherited balance + new overdraft_limit)

        if amount > available_funds:  # Checks against total available funds
            print("Overdraft limit exceeded! Transaction denied.")  # Error if exceeds overdraft
        else:
            self.balance -= amount  # Deducts from balance (inherited attribute)

            print(f"{amount} withdrawn by {self.name}. Balance: {self.balance}")  # Prints withdrawal (uses inherited name)

            if self.balance < 0:  # Checks if balance went negative (overdraft used)
                print("⚠ You are using overdraft facility")  # Warning for overdraft


#Savings account with interest feature
class SavingsAccount(Account):  # Subclass inherits from Account (hierarchical inheritance)
    def add_interest(self, rate):  # Adds new method specific to SavingsAccount (not in parent)
        interest = self.balance * (rate / 100)  # Calculates interest based on inherited balance
        self.balance += interest  # Updates inherited balance with interest
        print(f"Interest added at {rate}%. New balance: {self.balance}")  # Prints interest addition



#Premium account with perks and cashback feature
class PremiumAccount(Account):  # Subclass inherits from Account (hierarchical inheritance)
    def __init__(self, acc_no, name, balance, perks):  # Overrides __init__ to add perks
        super().__init__(acc_no, name, balance)  # Calls parent __init__ to set inherited attributes
        self.perks = perks  # Adds new attribute specific to PremiumAccount

    def show_perks(self):  # Adds new method specific to PremiumAccount
        print(f"{self.name}'s Premium Account Perks:")  # Prints perks (uses inherited name)
        for perk in self.perks:  # Loops through perks list
            print(f"- {perk}")  # Prints each perk

    def cashback(self, amount):  # Adds new method specific to PremiumAccount
        cashback_amount = amount * 0.02  # Calculates 2% cashback
        self.balance += cashback_amount  # Adds cashback to inherited balance
        print(f"Cashback of {cashback_amount} added. New balance: {self.balance}")  # Prints cashback addition



# Example of multilevel inheritance: Subclass of a subclass (REQUIRES MINIMUM DEPOSIT > 100,000)
class HighInterestSavingsAccount(SavingsAccount):  # Inherits from SavingsAccount (which inherits from Account)
    def __init__(self, acc_no, name, balance, high_rate_bonus=2):  # Overrides __init__ to add bonus and validation
        if balance <= 100000:  # Checks if minimum balance requirement is met
            raise ValueError(f"❌ Error: Amount should be greater than 100,000. You provided {balance}.")  # Raises error if balance too low
        else:  # If balance is valid (> 100,000)
            super().__init__(acc_no, name, balance)  # Calls SavingsAccount's __init__ (which calls Account's) — function continues
            self.high_rate_bonus = high_rate_bonus  # Adds new attribute — only runs if balance > 100,000

    def add_interest(self, rate):  # Overrides SavingsAccount's add_interest method
        total_rate = rate + self.high_rate_bonus  # Adds bonus to rate
        interest = self.balance * (total_rate / 100)  # Calculates interest with bonus
        self.balance += interest  # Updates balance
        print(f"High interest added at {total_rate}%. New balance: {self.balance}")  # Prints with bonus



# Creating an account object
#acc1
acc1 = Account("2183", "Aravind", 1000)  # Creates instance of base class Account
acc1.check_balance()  # Calls inherited method from Account

#acc2
acc2 = SavingsAccount("2184", "Anita", 2000)  # Creates instance of SavingsAccount (inherits from Account)
acc2.check_balance()  # Calls inherited method from Account
acc2.add_interest(5)  # Calls new method specific to SavingsAccount

#acc3
acc3 = PremiumAccount("2185", "Ravi", 5000, ["Priority Support", "Higher Withdrawal Limits"])  # Creates instance of PremiumAccount (inherits from Account)
acc3.check_balance()  # Calls inherited method from Account
acc3.show_perks()  # Calls new method specific to PremiumAccount
acc3.cashback(1000)  # Calls new method specific to PremiumAccount


#acc4
acc4 = CurrentAccount("2186", "Sita", 3000)  # Creates instance of CurrentAccount (inherits from Account)
acc4.check_balance()  # Calls inherited method from Account
acc4.withdraw(10000)  # Calls overridden method from CurrentAccount (polymorphism)

#acc5
acc5 = CurrentAccount("2187", "Kiran", 4000)  # Creates instance of CurrentAccount
acc5.check_balance()  # Calls inherited method

#acc6
acc6 = CurrentAccount("2188", "Maya", 5000)  # Creates instance of CurrentAccount
acc6.check_balance()  # Calls inherited method
acc6.withdraw(20000)  # Calls overridden method (tests overdraft limit)


accounts = [acc1, acc2, acc3, acc4, acc5, acc6]  # List of all account objects
for i in accounts:
    print(f"\nAccount: {i.name} | Type: {type(i).__name__} | Balance: {i.balance}")  # Prints account holder name, type of account, and balance


acc7 = HighInterestSavingsAccount("2189", "Vikram", 105000) 
acc7.check_balance()  # Calls inherited method from Account
acc7.add_interest(5)  # Calls overridden method from HighInterestSavingsAccount (includes bonus)
```

**Polymorphism**

```python

# =====================================================
# 🚗 1. RIDE SYSTEM (Polymorphism + Attributes)
# =====================================================

class Ride:
    def __init__(self, distance):
        # ATTRIBUTE: stores state of ride (distance traveled)
        self.distance = distance

    def calculate_fare(self):
        # This is a base method (to be overridden)
        raise NotImplementedError("Subclasses must implement this method")


class CarRide(Ride):
    def calculate_fare(self):
        # POLYMORPHISM: same method name, different behavior
        return self.distance * 15  # Car is expensive


class BikeRide(Ride):
    def calculate_fare(self):
        # POLYMORPHISM: Bike is cheaper
        return self.distance * 8


class AutoRide(Ride):
    def calculate_fare(self):
        # POLYMORPHISM: medium pricing
        return self.distance * 10


# =====================================================
# 💳 2. PAYMENT SYSTEM (Polymorphism + Attributes)
# =====================================================

class Payment:
    def __init__(self, amount):
        # ATTRIBUTE: stores payment amount
        self.amount = amount

    def pay(self):
        # Base method (must be overridden)
        raise NotImplementedError


class CardPayment(Payment):
    def pay(self):
        # POLYMORPHISM: Card payment behavior
        print(f"💳 Paid ₹{self.amount} using Credit/Debit Card")


class UPIPayment(Payment):
    def pay(self):
        # POLYMORPHISM: UPI behavior
        print(f"📱 Paid ₹{self.amount} using UPI")


class WalletPayment(Payment):
    def pay(self):
        # POLYMORPHISM: Wallet behavior
        print(f"💰 Paid ₹{self.amount} using Wallet")


# =====================================================
# 📢 3. NOTIFICATION SYSTEM (Polymorphism + Attributes)
# =====================================================

class Notification:
    def __init__(self, user):
        # ATTRIBUTE: stores user identity
        self.user = user

    def send(self, message):
        # Base method (must be overridden)
        raise NotImplementedError


class EmailNotification(Notification):
    def send(self, message):
        # POLYMORPHISM: email behavior
        print(f"📧 Email sent to {self.user}")
        print(f"Message: {message}")


class SMSNotification(Notification):
    def send(self, message):
        # POLYMORPHISM: SMS behavior
        print(f"📱 SMS sent to {self.user}")
        print(f"Message: {message}")


# =====================================================
# 🚀 4. MAIN APPLICATION FLOW
# =====================================================

# Step 1: Create a ride (object with attribute distance)
ride = BikeRide(distance=10)  
# Here:
# - object = BikeRide
# - attribute = distance = 10

# Step 2: Calculate fare (polymorphism in action)
fare = ride.calculate_fare()  
# Calls BikeRide.calculate_fare() automatically

print("\n--- RIDE COMPLETED ---")
print(f"Total Fare: ₹{fare}")

# Step 3: Payment selection (polymorphism)
payment = UPIPayment(amount=fare)  
# attribute: amount stored inside object

payment.pay()  
# Calls UPIPayment.pay()

print("\n--- PAYMENT COMPLETED ---")

# Step 4: Send notification (polymorphism)
notification = SMSNotification(user="Ravi")  
# attribute: user stored inside object

notification.send(f"Your ride of ₹{fare} is successful!")  
# Calls SMSNotification.send()

print("\n--- SYSTEM FINISHED ---")


ride2=CarRide(distance=20)
fare2=ride2.calculate_fare()
print(f"\nCar Ride Fare: ₹{fare2}")
payment2=CardPayment(amount=fare2)
payment2.pay()
notification2=EmailNotification(user="Anita")
notification2.send(f"Your car ride of ₹{fare2} is successful!")
```

**Encapsulation**
Wrapping data + restricting direct access + controlling it through methods

**Abstraction** Hide internal complexity and show only essential features

```python
# FULL UBER-LIKE OOP SYSTEM CODE

from abc import ABC, abstractmethod

# =====================================================
# 🎭 ABSTRACTION + ENCAPSULATION START
# =====================================================

class Vehicle(ABC):
    def __init__(self, distance):
        self.__distance = distance   # 🔒 ENCAPSULATION (hidden data)

    def get_distance(self):
        # controlled access to private data
        return self.__distance

    @abstractmethod
    def calculate_fare(self):
        # 🎭 ABSTRACTION: hiding implementation
        pass


# =====================================================
# 🧬 INHERITANCE + 🎭 POLYMORPHISM
# =====================================================

class Car(Vehicle):
    def calculate_fare(self):
        # POLYMORPHISM: custom logic for Car
        return self.get_distance() * 15


class Bike(Vehicle):
    def calculate_fare(self):
        # POLYMORPHISM: custom logic for Bike
        return self.get_distance() * 8


class Auto(Vehicle):
    def calculate_fare(self):
        # POLYMORPHISM: custom logic for Auto
        return self.get_distance() * 10


# =====================================================
# 🔒 ENCAPSULATION (Booking system hides details)
# =====================================================

class RideBooking:
    def __init__(self, vehicle, user):
        self.__vehicle = vehicle   # hidden object
        self.__user = user         # hidden user info

    def get_fare(self):
        # controlled access to vehicle logic
        return self.__vehicle.calculate_fare()

    def confirm_booking(self):
        print("====== RIDE BOOKED ======")
        print(f"User: {self.__user}")
        print(f"Distance: {self.__vehicle.get_distance()} km")
        print(f"Total Fare: ₹{self.get_fare()}")


# =====================================================
# 💳 PAYMENT SYSTEM (POLYMORPHISM)
# =====================================================

class Payment:
    def pay(self, amount):
        pass


class UPI(Payment):
    def pay(self, amount):
        print(f"\n📱 Paid ₹{amount} via UPI")


class Card(Payment):
    def pay(self, amount):
        print(f"\n💳 Paid ₹{amount} via Credit/Debit Card")


# =====================================================
# 🚀 MAIN FLOW (REAL SYSTEM)
# =====================================================

if __name__ == "__main__":

    # 1. User selects ride (INHERITANCE + POLYMORPHISM)
    vehicle = Bike(distance=12)

    # 2. Booking system (ENCAPSULATION)
    booking = RideBooking(vehicle, user="Ravi")
    booking.confirm_booking()

    # 3. Payment system (POLYMORPHISM)
    payment = UPI()
    payment.pay(booking.get_fare())

print("\n")
vehicle2 = Car(distance=25)
booking2 = RideBooking(vehicle2, user="Anita")
booking2.confirm_booking()

payment2 = Card()
payment2.pay(booking2.get_fare())
```
