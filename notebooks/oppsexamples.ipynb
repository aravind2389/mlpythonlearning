```python
class Dog:
    species = "Canine"  # Class attribute

    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age  # Instance attribute

# Creating an object of the Dog class
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)

print(dog1.name) 
print(dog1.species)
print(dog2.name)
print(dog2.age)
```

```python
# CLASS = Blueprint for creating bank accounts
class BankAccount:

    # __init__ is a special method (constructor)
    # It runs automatically when we create an object
    def __init__(self, name, balance):
        self.name = name        # Each attribute will have its own name
        self.balance = balance  # Each attribute will have its own balance

    # Method to deposit money
    def deposit(self, amount):
        self.balance += amount  # Add amount to balance
        print(f"{self.name} deposited {amount}")
        print(f"New balance: {self.balance}")

    # Method to withdraw money
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance ❌")
        else:
            self.balance -= amount
            print(f"{self.name} withdrew {amount}")
            print(f"Remaining balance: {self.balance}")

    # Method to check balance
    def check_balance(self):
        print(f"{self.name}'s current balance is {self.balance}")


# OBJECTS = Real bank accounts created from the class

# Creating first account
acc1 = BankAccount("Ravi", 1000)

# Creating second account
acc2 = BankAccount("Anita", 2000)


# USING OBJECTS

acc1.check_balance()
acc2.check_balance()

# Ravi deposits money
acc1.deposit(500)

# Anita withdraws money
acc2.withdraw(900)
acc2.withdraw(1100) 

# Check balances
acc1.check_balance()
acc2.check_balance()
```

```python
class ShoppingCart:
    def __init__(self, owner, items):
        self.owner = owner  # Owner of the shopping cart
        self.items = list((items))   # List of items in the cart

#Adding items to the cart
    def add_item(self, item):
        self.items.append(item)  # Add item to the cart
        print(f"Added {item} to the cart.")

#Removing an item from the cart
    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)  # Remove item from the cart
            print(f"Removed {item} from the cart.")
        else:
            print(f"{item} not found in the cart.")

    def view_cart(self):
        if not self.items:
            print("Your cart is empty.")
        else:
            print("Items in your cart:")
            for item in self.items:
                print(f"- {item}")


# Creating a shopping cart for Alice
cart1 = ShoppingCart("Alice", ["Apples", "Bananas"])

cart2 = ShoppingCart("Bob", ["Oranges", "Grapes"])

cart1.view_cart()  # View items in the cart
cart2.view_cart()  # View items in the cart

cart1.add_item("Milk")  # Add an item to Alice's cart
cart1.view_cart() # View updated cart for Alice

cart1.remove_item("Bananas")  # Remove an item from Alice's cart
cart1.view_cart() # View updated cart for Alice

cart1.remove_item("Bananas")
```

**Step-by-step thinking process (use this for ANY problem)**

Whenever you get a requirement, don’t jump into code. Break it like this:

**1. Identify “things” (→ Classes)**

Ask:

What are the main entities in this system?

For your e-commerce example:

Product
User
Cart
Order

👉 These become classes

**2. Identify “data” (→ Attributes)**

Ask:

What information does each thing store?

Example:

Product → name, price
User → name, cart, orders
Cart → items

👉 These become attributes

**3. Identify “actions” (→ Methods)**

Ask:

What can this thing DO?

Example:

Cart → add_item(), remove_item()
User → checkout()
Order → (mostly stores data)

👉 These become methods

**4. Identify relationships**

Ask:

How are these connected?

Example:

User has a Cart
Cart has many Products
User has many Orders

👉 This helps structure your code correctly

```python
# CLASS = Product → blueprint for products
class Product:

    def __init__(self, name, price):
        # ATTRIBUTES (data of each product object)
        self.name = name     # product name (e.g., Laptop)
        self.price = price   # product price (e.g., 50000)


# CLASS = ShoppingCart → blueprint for cart
class ShoppingCart:

    def __init__(self):
        # ATTRIBUTE → each cart has its own list of items
        self.items = []  
        # Example structure:
        # [{"product": product_object, "quantity": 2}]

    # METHOD → action (add item to cart)
    def add_item(self, product, quantity):

        # loop through existing items (THIS cart's data)
        for item in self.items:

            # check if product already exists
            if item["product"].name == product.name:

                # update quantity (modify attribute data)
                item["quantity"] += quantity
                print(f"Updated {product.name} quantity to {item['quantity']}")
                return

        # if product not found → add new item
        self.items.append({"product": product, "quantity": quantity})
        print(f"{product.name} x{quantity} added to cart")

    # METHOD → remove item
    def remove_item(self, product_name):
        for item in self.items:
            if item["product"].name == product_name:
                self.items.remove(item)
                print(f"{product_name} removed from cart")
                return
        print(f"{product_name} not found")

    # METHOD → update quantity
    def update_quantity(self, product_name, quantity):
        for item in self.items:
            if item["product"].name == product_name:

                # change attribute (quantity)
                item["quantity"] = quantity
                print(f"{product_name} quantity updated to {quantity}")
                return
        print(f"{product_name} not found")

    # METHOD → display cart
    def view_cart(self):
        print("\nCart items:")

        if not self.items:
            print("Cart is empty")
            return

        for item in self.items:
            p = item["product"]   # product object
            q = item["quantity"]  # quantity value

            # accessing product attributes
            print(f"- {p.name} | Price: {p.price} | Qty: {q}")

    # METHOD → calculate total price
    def calculate_total(self):
        total = 0

        # loop through cart items
        for item in self.items:
            # using product attributes + quantity
            total += item["product"].price * item["quantity"]

        return total


# CLASS = Order → represents a completed purchase
class Order:

    def __init__(self, items, total):
        # ATTRIBUTES
        self.items = items   # list of items purchased
        self.total = total   # total amount


# CLASS = User → represents a customer
class User:

    def __init__(self, name):
        # ATTRIBUTES
        self.name = name                 # user name
        self.cart = ShoppingCart()       # each user HAS a cart (object)
        self.orders = []                 # list to store past orders

    # METHOD → add product to cart
    def add_to_cart(self, product, quantity):
        print(f"\n{self.name} adding items...")

        # calling another object's method
        self.cart.add_item(product, quantity)

    # METHOD → apply discount
    def apply_discount(self, total):
        if total > 50000:
            print("Applying 10% discount")
            return total * 0.9
        return total

    # METHOD → checkout process
    def checkout(self):
        print(f"\n{self.name} is checking out...")

        # get total from cart object
        total = self.cart.calculate_total()

        # apply discount
        total = self.apply_discount(total)

        print(f"Final amount: {total}")

        # create ORDER OBJECT
        order = Order(self.cart.items.copy(), total)

        # store in user's order history
        self.orders.append(order)

        # clear cart after checkout
        self.cart.items.clear()

    # METHOD → view past orders
    def view_orders(self):
        print(f"\n{self.name}'s Orders:")

        if not self.orders:
            print("No orders yet")
            return

        for i, order in enumerate(self.orders, 1):
            print(f"\nOrder {i} - Total: {order.total}")

            for item in order.items:
                print(f"- {item['product'].name} x{item['quantity']}")


# ------------------ OBJECTS (REAL INSTANCES) ------------------

# Product objects
p1 = Product("Laptop", 50000)
p2 = Product("Phone", 20000)
p3 = Product("Headphones", 2000)

# User object
user = User("Ravi")


# ------------------ USING OBJECTS ------------------

user.add_to_cart(p1, 1)
user.add_to_cart(p3, 2)

user.add_to_cart(p3, 1)   # updates quantity

user.cart.view_cart()

user.cart.update_quantity("Laptop", 2)

user.cart.remove_item("Headphones")

user.cart.view_cart()

user.checkout()

user.view_orders()
```
